```python
import pandas as pd
import numpy as np

# Load the messy dataset[cite: 6]
df_messy = pd.read_csv('teaching_prices_messy.csv')

print("--- BEFORE CLEANING ---")
print(f"Shape: {df_messy.shape}")
print(df_messy.isnull().sum())
print(df_messy.dtypes)

# Always work on a copy[cite: 6]
df_clean = df_messy.copy()

# 1. Drop exact duplicate rows
df_clean = df_clean.drop_duplicates()

# 2. Clean string columns (strip whitespace and standardize case)
if 'item_name' in df_clean.columns:
    df_clean['item_name'] = df_clean['item_name'].astype(str).str.strip().str.title()
if 'item_category' in df_clean.columns:
    df_clean['item_category'] = df_clean['item_category'].astype(str).str.strip().str.lower()

# 3. Clean price column (remove currency symbols if present and convert to numeric)
if 'unit_price' in df_clean.columns:
    df_clean['unit_price'] = df_clean['unit_price'].astype(str).str.replace('$', '', regex=False).str.strip()
    df_clean['unit_price'] = pd.to_numeric(df_clean['unit_price'], errors='coerce')
    # Impute missing values with median
    df_clean['unit_price'] = df_clean['unit_price'].fillna(df_clean['unit_price'].median())

print("\n--- AFTER CLEANING ---")
print(f"Shape: {df_clean.shape}")
print(df_clean.isnull().sum())
print(df_clean.dtypes)

# Save cleaned output
df_clean.to_csv('teaching_prices_cleaned_output.csv', index=False)
```
